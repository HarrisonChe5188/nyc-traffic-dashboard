```python
#imports
import pandas as pd
```

```python
#load the data, prelimary exploration
df = pd.read_csv('North manhattan Before study 2026.csv')
print(df.head())
df.info()
df.describe()
```

```python
#convert the tt_median_timestamp column to datetime and extract the day of the week
df["tt_median_timestamp"] = pd.to_datetime(df["tt_median_timestamp"])
df["day_of_week"] = df["tt_median_timestamp"].dt.day_name()
print(df.info())
```

```python
#create a new column for the hour interval 
df["hour_interval"] = df["tt_median_timestamp"].dt.floor("h")
print(df.head())
```

```python
#save the cleaned dataframe to a new CSV file
df.to_csv("North_Manhattan_Cleaned.csv", index=False)
```
